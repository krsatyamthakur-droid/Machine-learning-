```python
# Real data

# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

# plot
plt.plot(years,oscar_revenue)
plt.title("oscar movie Revenue in each year")
plt.xlabel("years")
plt.ylabel("the revenue in million dollar")
plt.show()
```

```python
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]

years = [2008, 2009, 2010, 2011, 2012]

non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M
```

```python
plt.plot(years,oscar_revenue,label="Oscar movies")
plt.plot(years,non_oscar_revenue,label="non_oscar movie")
plt.xlabel("years")
plt.ylabel("the revenue in million dollar")
plt.title("oscar_movie revenue VS non_oscar_movie revenue")
plt.plot()
plt.legend()
```
