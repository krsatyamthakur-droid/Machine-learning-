### Q1.

A class has 4 students and 3 subjects. Their marks are given below. The examiner announces a subject-wise bonus of 5, 2 and 8 marks respectively. Add the bonus to every student in one line using broadcasting, then cap any mark above 100 back to 100 and report how many marks had to be capped.

```python
marks = np.array([[88, 96, 95],
                  [72, 99, 91],
                  [98, 85, 97],
                  [60, 75, 80]])
```

````python
# Q1
marks = np.array([[88, 96, 95],
                  [72, 99, 91],
                  [98, 85, 97],
                  [60, 75, 80]])

new_marks = marks + np.array([5, 2, 8])
print(new_marks)

capped = (new_marks > 100).sum()
new_marks = np.minimum(new_marks, 100)
print(new_marks)
print("marks capped:", capped)
````

### Q11.

Three sensors each recorded 5 readings. Stack them so that each sensor becomes a row, and separately stack them so that each sensor becomes a column. Print both shapes and confirm that one is the transpose of the other.

s1 = np.array([12, 15, 11, 19, 14])

### Q14.

Build a complete multi-step pipeline. Starting from the two arrays below: convert them into column vectors, join them into a single two-column table, attach a third column holding each row's total, then multiply the three-column table by `w = [0.5, 0.2, 0.3]` to obtain one score per row. Finally use `np.where` to mark each score as 1 if it is above the average score and 0 otherwise. Report the shape of every intermediate array.

```python
hours = np.array([2, 8, 5, 10, 3, 7])
attendance = np.array([55, 92, 70, 98, 60, 85])
```

````python
# Q14
hours = np.array([2, 8, 5, 10, 3, 7])
attendance = np.array([55, 92, 70, 98, 60, 85])

h = hours.reshape(-1, 1)
a = attendance.reshape(-1, 1)
print("column vectors:", h.shape, a.shape)

table = np.hstack((h, a))
print("two column table:", table.shape)

total = table.sum(axis=1).reshape(-1, 1)
table = np.hstack((table, total))
print("with total column:", table.shape)
print(table)

w = np.array([0.5, 0.2, 0.3])
score = table @ w
print("scores:", score, score.shape)

result = np.where(score > score.mean(), 1, 0)
print("average score:", score.mean())
print("above average:", result, result.shape)
````

### Q16.

Take the array below and create two 1D versions of it — one with `flatten()` and one with `ravel()`. Change the first element of each, then print the original matrix. Explain the difference you observe in a comment.

```python
m = np.array([[4, 8, 15],
              [16, 23, 42]])
```

````python
# Q16
m = np.array([[4, 8, 15],
              [16, 23, 42]])

f = m.flatten()
r = m.ravel()

f[0] = 100
print(m)

r[0] = 100
print(m)

# flatten() always makes a new copy, so changing f did not touch m.
# ravel() returns a view when it can, it shares memory with m,
# so changing r also changed m.
````
