# Part A — `tips.csv` (the restaurant)

A week of bills from one restaurant. Each row is one table: what they ate, what they paid,
what they tipped, which day, and how many people were sitting.

**A1.** The owner wants to know how generous each customer was, not just how much they left.

Make a new column called `tip_percent` — the tip as a **percentage of the bill**,
rounded to 1 decimal place. Then show only `total_bill`, `tip` and `tip_percent`
for the first 5 rows.

```python
# A1
df['tip_percent'] = (df['tip'] / df['total_bill'] * 100).round(1)
df[['total_bill', 'tip', 'tip_percent']].head()
```

**A2.** Answer both parts using **positions only** — no column names anywhere.

1. Print the `total_bill` of the **10th** row.
2. Show the **last 3 rows**, with only the **first two** columns.

```python
# A2
# 1. 10th row is position 9, total_bill is column 0
print(df.iloc[9, 0])

# 2.
df.iloc[-3:, :2]
```

**A3.** Now do it with **labels only**.

Show rows 20 to 25, with all the columns from `sex` through `size`.

Then answer in a comment: how many rows came back, and why is it not 5? ⭐

```python
# A3
df.loc[20:25, 'sex':'size']

# 6 rows came back, not 5.
# loc works with labels and includes the end label, so 25 is also included.
# (iloc[20:25] would stop before 25 and give 5)
```

**A4.** Two counts for the owner:

1. How many bills were **not** at Dinner? Use `~`.
2. How many bills had a party of **3 or more** *and* a tip **above 4**?

```python
# A4
# 1.
not_dinner = df[~(df['time'] == 'Dinner')]
print("not at dinner:", len(not_dinner))

# 2.
big_tip = df[(df['size'] >= 3) & (df['tip'] > 4)]
print("party of 3+ and tip above 4:", len(big_tip))
```

**A5.** The owner wants to flag the generous customers.

Make a column called `service`. Everyone starts as `'normal'`.
Any bill where `tip_percent` is **more than 20** becomes `'generous'`.

Then count how many of each. (This uses the column you made in A1.)

```python
# A5
df['service'] = 'normal'
df.loc[df['tip_percent'] > 20, 'service'] = 'generous'

df['service'].value_counts()
```

# Part B — `Titanic-Dataset.csv`

⚠️ Remember: the column names here start with **capital letters** —
`Survived`, `Pclass`, `Sex`, `Age`, `SibSp`, `Parch`, `Fare`, `Cabin`, `Embarked`.

**B1.** How many passengers boarded from each port? Use the `Embarked` column.

Then add the three numbers up in a comment. Do they come to 891? ⭐

```python
# B1
print(t['Embarked'].value_counts())

# 644 + 168 + 77 = 889
# no, it does not come to 891. 2 passengers have no Embarked value (NaN),
# and value_counts() skips NaN by default.
```

**B2.** Two counts:

1. How many **women** were travelling in **third class**?
2. How many of those women **survived**?

```python
# B2
women_3rd = t[(t['Sex'] == 'female') & (t['Pclass'] == 3)]
print("women in third class:", len(women_3rd))

survived = women_3rd[women_3rd['Survived'] == 1]
print("of them survived:", len(survived))
```

**B3.** Some people paid an enormous amount for their ticket.

Show only `Name`, `Age` and `Fare` for every passenger who paid **more than 200**.
How many are there?

```python
# B3
rich = t.loc[t['Fare'] > 200, ['Name', 'Age', 'Fare']]
print("passengers who paid more than 200:", len(rich))
```

**B4.** Out of all the passengers in **first class**, what percentage survived?

Now do the same for **third class**. Put the two numbers next to each other in a comment.

```python
# B4
first = t[t['Pclass'] == 1]
third = t[t['Pclass'] == 3]

first_pct = len(first[first['Survived'] == 1]) / len(first) * 100
third_pct = len(third[third['Survived'] == 1]) / len(third) * 100

print("first class survived:", round(first_pct, 2), "%")
print("third class survived:", round(third_pct, 2), "%")

# first class: 62.96 %   vs   third class: 24.24 %
```

**B5.** Make a column called `age_group`. Everyone starts as `'adult'`.
Anyone under 18 becomes `'child'`. Then count how many of each.

**Think before tomorrow (no code needed).** 177 passengers have no `Age` recorded at all.
Which group did they land in — and is that the right place for them?
Write two lines in a comment saying what you would do about it. ⭐

```python
# B5
t['age_group'] = 'adult'
t.loc[t['Age'] < 18, 'age_group'] = 'child'

print(t['age_group'].value_counts())

# the 177 passengers with no Age all landed in 'adult', because NaN < 18 is False
# so they just kept the starting value. that is not right, some of them could be kids.
# i would give them a separate 'unknown' group, or fill the age first
# (for example with the median age) before making the groups.
```
