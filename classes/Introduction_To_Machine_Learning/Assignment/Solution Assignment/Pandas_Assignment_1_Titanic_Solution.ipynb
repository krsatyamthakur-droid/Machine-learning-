**Q8.** In `describe()`, compare the mean of `Age` with its 50% value. Which is larger, and what does that suggest? ★

```python
print("mean:", round(t['Age'].mean(), 2))
print("50% (median):", t['Age'].median())

# mean (29.7) is larger than the median (28).
# most passengers were young adults, but a smaller group of older people (up to 80)
# pulls the mean up. so age is slightly right skewed.
```

**Q9.** (Challenge) Using only `shape` and `info()`, work out what percentage of age values are missing.

```python
total_rows = t.shape[0]    # 891
age_present = 714          # from info()
missing = total_rows - age_present

print(missing)
print(round(missing / total_rows * 100, 2), "%")
# about 19.87 % of age values are missing
```
