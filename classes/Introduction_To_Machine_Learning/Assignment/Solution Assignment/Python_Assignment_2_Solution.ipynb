## Part A · Strings

**Q1.** A user types their city carelessly: `"   nEw   dELhi  "`. Clean it and print exactly `New Delhi`. Also validate that the city contains only alphabets and spaces — print `Invalid city` otherwise.

```python
city = "   nEw   dELhi  "

city = " ".join(city.split()).title()

if city.replace(" ", "").isalpha():
    print(city)
else:
    print("Invalid city")
```

**Q2.** Prices arrive as one messy string: `" 120, 45 , 300 ,80, 95 "`. Convert it into a list of integers, then print the total bill, the highest price, and the average price with 2 decimal places.

```python
prices_str = " 120, 45 , 300 ,80, 95 "

prices = []
for p in prices_str.split(","):
    prices.append(int(p.strip()))

total = sum(prices)
highest = max(prices)
average = total / len(prices)

print(f"Total   : {total}")
print(f"Highest : {highest}")
print(f"Average : {average:.2f}")
```

**Q3.** Validate a username (e.g. `"   Rahul_123  "`). Clean spaces and lowercase it, it must be alphanumeric, must start with an alphabet, and length must be between 5 and 12.

```python
def check_username(username):
    username = username.strip().lower()

    if (username.isalnum()
            and username[0].isalpha()
            and 5 <= len(username) <= 12):
        return "Valid username"
    return "Invalid username"

print(check_username("   Rahul_123  "))
print(check_username("rahul123"))
print(check_username("1rahul"))
```

**Q4.** For a given sentence, print: (a) the total number of words, (b) how many words end with a vowel, and (c) the longest word.

```python
sentence = "Python is an amazing language to explore and learn"

words = sentence.split()
print("Total words:", len(words))

vowel_count = 0
for word in words:
    if word[-1].lower() in "aeiou":
        vowel_count += 1
print("Words ending with a vowel:", vowel_count)

longest = max(words, key=len)
print("Longest word:", longest)
```

## Part B · Dictionaries

**Q5.** Count the frequency of each character in a string, ignoring spaces and treating uppercase and lowercase as the same.

```python
text = "Aba c"

freq = {}
for ch in text.lower():
    if ch == " ":
        continue
    freq[ch] = freq.get(ch, 0) + 1

print(freq)
```

**Q6.** Count the frequency of each word in a sentence and print the word that occurs the most.

```python
sentence = "the cat and the dog and the bird"

word_count = {}
for word in sentence.split():
    word_count[word] = word_count.get(word, 0) + 1

print(word_count)

sorted_words = sorted(word_count.items(), key=lambda x: x[1], reverse=True)
top_word, times = sorted_words[0]
print(f"Most frequent word: '{top_word}' ({times} times)")
```

**Q7.** Invert the dictionary so that values become keys. If two days share the same value, store both days in a list.

```python
sales = {"mon": 200, "tue": 350, "wed": 200}

inverted = {}
for day, amount in sales.items():
    if amount not in inverted:
        inverted[amount] = []
    inverted[amount].append(day)

print(inverted)
```

**Q8.** Merge the marks of two tests into one dictionary — if a student appears in both, sum the marks.

```python
test1 = {"amit": 40, "bina": 35}
test2 = {"amit": 30, "chirag": 45}

total = test1.copy()
for name, marks in test2.items():
    total[name] = total.get(name, 0) + marks

print(total)
```

## Part C · Sorting + JSON-style Data

**Q9.** Sort the employees by salary (highest first) using `sorted()` with `key=lambda`, and print each line with an f-string.

```python
employees = [("Ravi", 55000), ("Sita", 72000), ("Amit", 48000)]

for name, salary in sorted(employees, key=lambda emp: emp[1], reverse=True):
    print(f"{name} earns Rs. {salary}")
```

**Q10.** Using the list of movie dictionaries: (a) titles with rating > 8.1, (b) all genres in a list without duplicates, (c) a `{rating: [titles]}` dict printed rating by rating.

```python
movies = [
    {"title": "Movie A", "rating": 8.2, "genre": "Thriller"},
    {"title": "Movie B", "rating": 8.4, "genre": "Sports"},
    {"title": "Movie C", "rating": 8.1, "genre": "Comedy"},
    {"title": "Movie D", "rating": 8.2, "genre": "Drama"},
]

# (a)
print("Movies rated above 8.1:")
for movie in movies:
    if movie["rating"] > 8.1:
        print(" -", movie["title"])

# (b)
genres = []
for movie in movies:
    if movie["genre"] not in genres:
        genres.append(movie["genre"])
print("\nGenres:", genres)

# (c)
by_rating = {}
for movie in movies:
    r = movie["rating"]
    if r not in by_rating:
        by_rating[r] = []
    by_rating[r].append(movie["title"])

print("\nMovies grouped by rating:")
for rating in sorted(by_rating, reverse=True):
    print(f"  Rating {rating} : {', '.join(by_rating[rating])}")
```
